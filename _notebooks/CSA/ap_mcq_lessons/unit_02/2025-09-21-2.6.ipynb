# Quick Recap: Boolean Expressions

A **boolean expression** evaluates to exactly one of two values: `true` or `false`. You build them with:

- **Relational operators**: `==`, `!=`, `<`, `>`, `<=`, `>=`
- **Logical operators**: `&&` (AND), `\|\|` (OR), `!` (NOT)

```python
public class BooleanBasics {
    public static void main(String[] args) {
        int age = 16;
        boolean hasPermit = true;

        boolean canDriveAlone = (age >= 16) && hasPermit;
        boolean isTeen = (age >= 13) && (age <= 19);

        System.out.println("canDriveAlone: " + canDriveAlone);
        System.out.println("isTeen: " + isTeen);
    }
}

BooleanBasics.main(null);
```

Think of two boolean expressions like two light switches wired differently in the wall. They might look nothing alike, but if they always turn the same light on and off together, no matter how you flip them, they're doing the **same job**. That's what it means for boolean expressions to be **equivalent**.

# Comparing Boolean Expressions

Two boolean expressions are **equivalent** if they produce the *same result* for **every possible combination** of their input values, not just some of them.

The tool we use to check this is a **truth table**: a chart that lists every possible T/F combination of the variables involved, and the resulting value of each expression.

## Truth Table Example

Is `!(a && b)` equivalent to `!a || !b`?

| a | b | a && b | !(a && b) | !a | !b | !a \|\| !b |
|---|---|--------|-----------|----|----|-----------|
| T | T | T | F | F | F | F |
| T | F | F | T | F | T | T |
| F | T | F | T | T | F | T |
| F | F | F | T | T | T | T |

The last two columns match on **every row**, so the two expressions are equivalent!

## De Morgan's Laws

These two laws tell you how to "distribute" a `!` across a compound boolean expression:

- `!(a && b)` is equivalent to `!a || !b`
- `!(a || b)` is equivalent to `!a && !b`

**Trick to remember it:** when the `!` moves inside the parentheses, flip `&&` to `\|\|` (or `\|\|` to `&&`), and negate each individual piece.

```python
public class DeMorganDemo {
    public static void main(String[] args) {
        // Testing every combination of a and b
        boolean[] values = {true, false};

        for (boolean a : values) {
            for (boolean b : values) {
                boolean left  = !(a && b);
                boolean right = !a || !b;
                System.out.println("a=" + a + ", b=" + b +
                    " -> !(a&&b)=" + left + ", !a||!b=" + right +
                    ", equivalent? " + (left == right));
            }
        }
    }
}

DeMorganDemo.main(null);
```

**College Board tests these concepts!**

- Equivalent means true for **every** possible input, not just one you tried.
- `!(a && b)` is **not** the same as `!a && !b`. Only the De Morgan pairing above is equivalent.
- Precedence order for evaluation: `!` first, then `&&`, then `\|\|`.
- The AP exam loves giving you a messy negated expression and asking which of 4 choices is equivalent to it.

# Short-Circuit Evaluation

Java doesn't always evaluate the whole boolean expression. It stops as soon as it already knows the final answer:

- For `&&`: if the **left** side is `false`, the whole thing must be `false` — Java never checks the right side.
- For `\|\|`: if the **left** side is `true`, the whole thing must be `true` — Java never checks the right side.

Think of it like a bouncer checking IDs at a door with two rules joined by AND: the moment you fail rule one, the bouncer stops reading rule two, you're not getting in either way.

```python
public class ShortCircuitDemo {

    public static boolean noisyCheck(String label, boolean value) {
        System.out.println("Checking: " + label);
        return value;
    }

    public static void main(String[] args) {
        System.out.println("--- AND test ---");
        boolean andResult = noisyCheck("left (false)", false) && noisyCheck("right", true);
        System.out.println("Result: " + andResult);

        System.out.println("--- OR test ---");
        boolean orResult = noisyCheck("left (true)", true) || noisyCheck("right", false);
        System.out.println("Result: " + orResult);
    }
}

ShortCircuitDemo.main(null);
```

Notice that `"right"` is never printed in either test. This isn't just a quirk, it's genuinely useful. A classic example is guarding against a crash:

```python
public class GuardExample {
    public static void main(String[] args) {
        String name = null;

        // Safe: short-circuit stops before name.length() runs
        if (name != null && name.length() > 0) {
            System.out.println("Name has letters");
        } else {
            System.out.println("Name is null or empty, safely handled");
        }
    }
}

GuardExample.main(null);
```

# Popcorn Hack #2: Are These Equivalent?

Examine each pair of expressions below and predict, without running anything, whether they are equivalent.

```python
public class EquivalenceCheck {
    public static void main(String[] args) {
        boolean[] values = {true, false};

        System.out.println("Pair 1: !(a || b)  vs  !a && !b");
        for (boolean a : values) {
            for (boolean b : values) {
                boolean left = !(a || b);
                boolean right = !a && !b;
                System.out.println("a=" + a + ", b=" + b + " -> " + left + " | " + right);
            }
        }

        System.out.println("\nPair 2: a && (b || c)  vs  (a && b) || c");
        for (boolean a : values) {
            for (boolean b : values) {
                for (boolean c : values) {
                    boolean left = a && (b || c);
                    boolean right = (a && b) || c;
                    System.out.println("a=" + a + ", b=" + b + ", c=" + c +
                        " -> " + left + " | " + right);
                }
            }
        }
    }
}

EquivalenceCheck.main(null);
```

**Questions:**

1. Is Pair 1 equivalent for every row? Which law does this confirm?
2. Is Pair 2 equivalent for every row? Find one combination of `a`, `b`, `c` where the two sides disagree.
3. Why is it not enough to check just one or two rows before calling two expressions "equivalent"?

# Key Takeaways

- A boolean expression only ever evaluates to `true` or `false`
- Two expressions are equivalent only if they match on **every row** of the truth table
- De Morgan's Laws: `!(a && b)` = `!a || !b`, and `!(a || b)` = `!a && !b`
- `&&` and `\|\|` short-circuit: Java skips the right side once the result is already determined
- Short-circuit evaluation is often used to safely guard against errors (like null values)
- These concepts are frequently tested on the College Board AP CSA exam!
